```python
import os
import pandas as pd

# --- CONFIGURATION ---
XLSX_PATH = "/content/drive/MyDrive/dem_fluency.xlsx"  # change to your xlsx path
OUTPUT_ROOT = "/content/drive/MyDrive/dementia/recall/processed_audios"

# --- LOAD XLSX ---
df = pd.read_excel(XLSX_PATH, header=0)
xlsx_ids = set(df.iloc[:, 0].astype(str).str.strip())  # first column
# include the header value itself since it's also an audio ID
xlsx_ids.add(df.columns[0].strip())

print(f"📊 Total audio IDs in Excel: {len(xlsx_ids)}")

# --- GET PROCESSED FOLDERS ---
if os.path.exists(OUTPUT_ROOT):
    processed_folders = set(os.listdir(OUTPUT_ROOT))
else:
    processed_folders = set()

print(f"📁 Total folders in processed_audios: {len(processed_folders)}")

# --- CROSS CHECK ---
in_excel_not_folder = xlsx_ids - processed_folders
in_folder_not_excel = processed_folders - xlsx_ids
matching = xlsx_ids & processed_folders

print(f"\n✅ Matching (in both):            {len(matching)}")
print(f"❌ In Excel but NO folder:        {len(in_excel_not_folder)}")
print(f"⚠️  Folder exists but NOT in Excel: {len(in_folder_not_excel)}")

if in_excel_not_folder:
    print(f"\n--- In Excel but missing folder ---")
    for name in sorted(in_excel_not_folder):
        print(f"  {name}")

if in_folder_not_excel:
    print(f"\n--- Folder exists but not in Excel ---")
    for name in sorted(in_folder_not_excel):
        print(f"  {name}")

# --- CHECK EMPTY FOLDERS OR MISSING FILES ---
print(f"\n--- File check per folder ---")
empty_folders = []
missing_files_report = []

for folder_name in sorted(processed_folders):
    folder_path = os.path.join(OUTPUT_ROOT, folder_name)
    if not os.path.isdir(folder_path):
        continue

    files = os.listdir(folder_path)
    if not files:
        empty_folders.append(folder_name)
        continue

    missing = []
    if not any(f.endswith(".mp3") for f in files):
        missing.append("audio.mp3")
    if "diarization.rttm" not in files:
        missing.append("diarization.rttm")
    if "exclusive_diarization.rttm" not in files:
        missing.append("exclusive_diarization.rttm")
    if "turn_level.csv" not in files:
        missing.append("turn_level.csv")

    if missing:
        missing_files_report.append((folder_name, missing))

print(f"\n🗂️  Empty folders: {len(empty_folders)}")
for name in sorted(empty_folders):
    print(f"  {name}")

print(f"\n📂 Folders with missing files: {len(missing_files_report)}")
for folder_name, missing in sorted(missing_files_report):
    print(f"  {folder_name}: missing → {', '.join(missing)}")

print("\n--- CROSS CHECK COMPLETE ---")
```
