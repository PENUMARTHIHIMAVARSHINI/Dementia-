```python
# Step 2: Import libraries
import pandas as pd
import os
```

```python
# Step 3: Read Excel (no header)
excel_path = "/content/drive/MyDrive/cont_fluency.xlsx"

df = pd.read_excel(excel_path, header=None)

# 👉 Assuming your values are in FIRST column
excel_values = df[0].astype(str).str.strip()

print("Excel sample:")
print(excel_values.head())
```

```python
# Step 4: Read ALL folder names inside 'cookie'
base_path = "/content/drive/MyDrive/control/fluency/processed_audios"

folder_names = [
    name.strip()
    for name in os.listdir(base_path)
    if os.path.isdir(os.path.join(base_path, name))
]

print("Folder sample:")
print(folder_names[:5])
```

```python
# Step 5: Normalize (IMPORTANT for matching)
excel_values = excel_values.str.lower()
folder_names = [f.lower() for f in folder_names]
```

```python
# Step 6: Find missing (Excel but NOT in folders)
missing_in_drive = sorted(set(excel_values) - set(folder_names))

print("Missing entries:")
print(missing_in_drive)
```
